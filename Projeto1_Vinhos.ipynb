# Projeto 1 - Analisando Vinhos

```python
import numpy as np
from sklearn import datasets
import matplotlib.pyplot as plt

# Baixar a base de dados wine
dados = datasets.load_wine()
print("Descricao da base de dados :: ", dados['DESCR'])

#Obter os dados
X = dados.data
Y = dados.target
```

```python
# Visualizar a dimensao da matriz de dados e do vetor de rotulos
print('Dimensão de X =', X.shape)
print('Dimensão de Y =', Y.shape)
print('Nome das classes:', dados.target_names)
print('Nome das características:', dados.feature_names)
```

```python
# Contar quantas amostras existem para cada classe de vinho
vinho1 = Y == 0
num_vinho1 = sum(vinho1)

vinho2 = Y == 1
num_vinho2 = sum(vinho2)

vinho3 = Y == 2
num_vinho3 = sum(vinho3)

print('num. vinho classe 0:', num_vinho1)
print('num. vinho classe 1:', num_vinho2)
print('num. vinho classe 2:', num_vinho3)
```

Valores dentro de cada classe de vinho

```python
# Separando os dados de cada classe de vinho considerando os limites de cada classe
# Classe 0: 0 ate 59 (59 amostras)
# Classe 1: 59 ate 130 (71 amostras)
# Classe 2: 130 ate 178 (48 amostras)
x_vinho1 = X[0:59, :]
x_vinho2 = X[59:130, :]
x_vinho3 = X[130:178, :]

print(x_vinho1.shape)
print(x_vinho2.shape)
print(x_vinho3.shape)
```

```python
# Encontrar maximo, minimo, media e desvio padrao para cada variavel em cada uma das tres classes
print("Vinho Classe 0")
max_v1 = np.max(x_vinho1, axis=0)
print('Máximo =', max_v1)
min_v1 = np.min(x_vinho1, axis=0)
print('Mínimo =', min_v1)
media_v1 = np.mean(x_vinho1, axis=0)
print('média:', media_v1)
desvio_v1 = np.std(x_vinho1, axis=0)
print('desvio_padrão:', desvio_v1)

print("\nVinho Classe 1")
max_v2 = np.max(x_vinho2, axis=0)
print('Máximo =', max_v2)
min_v2 = np.min(x_vinho2, axis=0)
print('Mínimo =', min_v2)
media_v2 = np.mean(x_vinho2, axis=0)
print('média:', media_v2)
desvio_v2 = np.std(x_vinho2, axis=0)
print('desvio_padrão:', desvio_v2)

print("\nVinho Classe 2")
max_v3 = np.max(x_vinho3, axis=0)
print('Máximo =', max_v3)
min_v3 = np.min(x_vinho3, axis=0)
print('Mínimo =', min_v3)
media_v3 = np.mean(x_vinho3, axis=0)
print('média:', media_v3)
desvio_v3 = np.std(x_vinho3, axis=0)
print('desvio_padrão:', desvio_v3)
```

```python
# Plotar os dados considerando Flavanoids (coluna 6) no eixo x e Proline (coluna 12) no eixo y
plt.scatter(x_vinho1[:, 6], x_vinho1[:, 12],
            color='green', marker='o', label='classe 0')
plt.scatter(x_vinho2[:, 6], x_vinho2[:, 12],
            color='red', marker='o', label='classe 1')
plt.scatter(x_vinho3[:, 6], x_vinho3[:, 12],
            color='blue', marker='o', label='classe 2')

plt.title('Valores de Flavanoids e Proline para cada classe de vinho')
plt.xlabel('Flavanoids')
plt.ylabel('Proline')
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

```python
# Plotar os tres pontos correspondentes as medias de Flavanoids (indice 6) e Proline (indice 12)
ponto_v1 = np.array([media_v1[6], media_v1[12]])
ponto_v2 = np.array([media_v2[6], media_v2[12]])
ponto_v3 = np.array([media_v3[6], media_v3[12]])

plt.scatter(ponto_v1[0], ponto_v1[1],
            color='green', marker='o', label='média classe 0')
plt.scatter(ponto_v2[0], ponto_v2[1],
            color='red', marker='o', label='média classe 1')
plt.scatter(ponto_v3[0], ponto_v3[1],
            color='blue', marker='o', label='média classe 2')

plt.title('Valores médios de Flavanoids e Proline')
plt.xlabel('Flavanoids')
plt.ylabel('Proline')
plt.legend(loc='upper left')
plt.grid()
plt.show()

# Calcular as distancias Euclidianas entre os pontos medios
distancia_v1_v2 = np.sqrt(sum((ponto_v1 - ponto_v2) ** 2))
distancia_v1_v3 = np.sqrt(sum((ponto_v1 - ponto_v3) ** 2))
distancia_v2_v3 = np.sqrt(sum((ponto_v2 - ponto_v3) ** 2))

print('Distância classe 0 - classe 1: ', format(distancia_v1_v2, ".2f"))
print('Distância classe 0 - classe 2: ', format(distancia_v1_v3, ".2f"))
print('Distância classe 1 - classe 2: ', format(distancia_v2_v3, ".2f"))
```
